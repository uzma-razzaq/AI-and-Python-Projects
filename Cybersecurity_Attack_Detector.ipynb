```python
from sklearn.tree import DecisionTreeClassifier
import numpy as np

# 1. Data: (Data Packet ka Size KB mein, Login Attempts ki Tadaad)
network_traffic = np.array(( (20, 1), (35, 2), (500, 15), (800, 25) ))

# Result: 0 = Safe (Normal User), 1 = Danger (Hacker Attack)
status = np.array(( 0, 0, 1, 1 ))

# 2. AI Security Model banana aur train karna
security_ai = DecisionTreeClassifier()
security_ai.fit(network_traffic, status)

# 3. Naya Test: Agar kisi ne 700 KB data bheja aur 18 baar login try kiya
new_activity = np.array(( (700, 18), ))
prediction = security_ai.predict(new_activity)

if prediction.item() == 1:
    print("Alert: Yeh Hacker Attack (Anomaly) hai!")
else:
    print("Safe: Yeh Normal User hai.")
```
