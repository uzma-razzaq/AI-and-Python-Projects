```python
import streamlit as st
from sklearn.ensemble import RandomForestClassifier
import numpy as np

# 1. Website ka Design (Frontend)
st.title("🛡️ AI Phishing Link Detector")
st.write("Developed by Uzma Razzaq | Cybersecurity AI Tool")
st.write("Neeche koi bhi website link paste karein aur check karein ke woh Safe hai ya Hacker ka link:")

# 2. Chota AI Model Train Karna (Backend Dimagh)
# Features: [Link ki Lambai, Dots ki Tadaad, Kya '@' ka nishan hai (1=Yes, 0=No)]
X_train = np.array((
    (18, 1, 0),   # Safe: google.com
    (22, 2, 0),   # Safe: www.wikipedia.org
    (75, 5, 1),   # Danger: lamba link + @ sign
    (90, 6, 1),   # Danger: bohot lamba + zyada dots
    (25, 1, 0),   # Safe: github.com
    (82, 4, 1)    # Danger: phishing link
))
y_train = np.array((0, 0, 1, 1, 0, 1)) # 0 = Safe, 1 = Phishing (Danger)

model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# 3. User se Link Lena aur Result Dikhana
url = st.text_input("Website Link Yahan Likhein (e.g., https://google.com):")

if st.button("Check Security Now"):
    if url:
        # Link ki khasoosiyat nikalna
        length = len(url)
        dots = url.count('.')
        has_at = 1 if '@' in url else 0

        # AI se faisla karwana
        features = np.array(((length, dots, has_at),))
        prediction = model.predict(features)

        if prediction.item() == 1:
            st.error("⚠️ DANGER: Yeh Phishing (Hacker) Link ho sakta hai! Is par click mat karein.")
        else:
            st.success("✅ SAFE: Yeh website link bilkul mehfooz lag raha hai.")
    else:
        st.warning("Pehle upar box mein koi link paste karein!")
```
