```python
from sklearn.linear_model import LinearRegression
import numpy as np

# 1. Data: Ghar ka size (Marla mein) aur uski Keemat (Lakh mein)
size = np.array([[5], [10], [15], [20]])
price = np.array([50, 100, 150, 200])

# 2. AI Model banana aur sikhana (Training)
model = LinearRegression()
model.fit(size, price)

# 3. Naye ghar ki keemat predict karna (Farz karein 12 Marla ka ghar hai)
predicted_price = model.predict([[25]])
print("25 Marla ke ghar ki andazan keemat (Lakh mein):", predicted_price[0])
```

```python
from sklearn.ensemble import RandomForestRegressor
import numpy as np

# 1. Data: (Router se Faasla meters mein, Signal Strength 1 se 100 tak)
wifi_data = np.array(( (2, 95), (5, 80), (10, 60), (20, 30) ))

# Internet Speed (Mbps mein)
internet_speed = np.array(( 100, 75, 45, 15 ))

# 2. AI Model banana (Random Forest)
ai_model = RandomForestRegressor(n_estimators=10, random_state=42)
ai_model.fit(wifi_data, internet_speed)

# 3. Prediction: Agar router se 7 meter door hon aur signal 75 ho, toh speed kya hogi?
new_test = np.array(( (7, 75), ))
predicted_speed = ai_model.predict(new_test)

print("Andazan Internet Speed (Mbps):", round(predicted_speed.item(), 2))
```
